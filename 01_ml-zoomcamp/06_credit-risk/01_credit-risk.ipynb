```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('scores.csv')
df.head().T
```

```python
df.shape
```

```python
df.columns = df.columns.str.lower()
```

```python
df.dtypes
```

```python
df.status.value_counts()
```

```python
status_values = {
    1: 'ok',
    2: 'default',
    0: 'unknown'
}

df.status = df.status.map(status_values)
```

```python
home_values = {
    1: 'rent',
    2: 'owner',
    3: 'private',
    4: 'ignore',
    5: 'parents',
    6: 'other',
    0: 'unk'
}

df.home = df.home.map(home_values)
```

```python
marital_values = {
    1: 'single',
    2: 'married',
    3: 'widow',
    4: 'separated',
    5: 'divorced',
    0: 'unk'
}

df.marital = df.marital.map(marital_values)
```

```python
records_values = {
    1: 'no',
    2: 'yes',
    0: 'unk'
}

df.records = df.records.map(records_values)
```

```python
job_values = {
    1: 'fixed',
    2: 'partime',
    3: 'freelance',
    4: 'others',
    0: 'unk'
}

df.job = df.job.map(job_values)
```

```python
df.head()
```

```python
df.dtypes
```

```python
df.describe().round(2)
```

```python
for col in ['income', 'assets', 'debt']:
    df[col] = df[col].replace(99999999, np.nan)
```

```python
df.describe().round(2)
```

```python
df = df.drop(df[df.status == 'unknown'].index)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.fillna(0)
```

```python
from sklearn.model_selection import train_test_split

train_full_df, test_df = train_test_split(df, test_size=0.2, random_state=1)
train_df, val_df = train_test_split(train_full_df, test_size=0.25, random_state=1)
```

```python
train_full_df.shape, train_df.shape, val_df.shape, test_df.shape
```

```python
train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

y_train = (train_df.status == 'ok').astype(int).values
y_val = (val_df.status == 'ok').astype(int).values
y_test = (test_df.status == 'ok').astype(int).values
```

```python
del train_df['status']
del val_df['status']
del test_df['status']
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction import DictVectorizer
```

```python
dv = DictVectorizer(sparse=False)

X_train = dv.fit_transform(train_df.to_dict(orient='records'))
X_val = dv.transform(val_df.to_dict(orient='records'))
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
y_pred_train = model.predict(X_train)
y_pred_val = model.predict(X_val)
```

```python
from sklearn.metrics import roc_auc_score

print('train auc-roc score', roc_auc_score(y_train, y_pred_train))
print('validation auc-roc score', roc_auc_score(y_val, y_pred_val))
```

```python
improved_model = DecisionTreeClassifier(max_depth=2)
improved_model.fit(X_train, y_train)

y_pred_train = improved_model.predict(X_train)
y_pred_val = improved_model.predict(X_val)

print('train auc-roc score', roc_auc_score(y_train, y_pred_train))
print('validation auc-roc score', roc_auc_score(y_val, y_pred_val))
```

```python
stump_model = DecisionTreeClassifier(max_depth=1)
stump_model.fit(X_train, y_train)

y_pred_train = stump_model.predict(X_train)
y_pred_val = stump_model.predict(X_val)

print('train auc-roc score', roc_auc_score(y_train, y_pred_train))
print('validation auc-roc score', roc_auc_score(y_val, y_pred_val))
```

```python
from sklearn.tree import export_text

print(export_text(improved_model))
```

```python
data = [
    [8000, 'default'],
    [2000, 'default'],
    [0, 'default'],
    [5000, 'ok'],
    [5000, 'ok'],
    [4000, 'ok'],
    [9000, 'ok'],
    [3000, 'default']
    ]

example_df = pd.DataFrame(data, columns=['assets', 'status'])
example_df
```
